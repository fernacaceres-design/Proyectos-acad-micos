# Obtención de datos

## Web Scraper

```python
# Configuración del Webdriver
chrome_options = webdriver.ChromeOptions()
chrome_options.add_argument('--headless')
chrome_options.add_argument('--no-sandbox')
chrome_options.add_argument('--disable-dev-shm-usage')
```

```python
# Función get_links
# Scrapea todos los links dentro de una página
def get_links(url, url_base, url_data):
    global chrome_options
    print(f"Scrapeando links desde {url}")

    # Inicializamos el navegador
    driver = webdriver.Chrome(options=chrome_options)
    driver.get(url)

    # Obtenemos todos los elementos <a> de la página (hipervínculos)
    links = driver.find_elements(By.TAG_NAME, 'a')

    i = 0
    for l in links:
        try:
            # Obtenemos el enlace
            link = ":".join(l.get_attribute('href').split("#")[0].split(":")[:2])
            if url_base in link and link not in url_data and "mediawiki" not in link:
                url_data.append(link)

        except:
            pass

        print(f"\r\tLinks revisados: {i}/{len(links)}", end="")
        i += 1

    print(f"\r\tLinks revisados: {i}/{len(links)}")

    driver.quit()
```

```python
# Función get_text
# Scrapea el texto dentro de una página web
def get_text(url):
    global chrome_options

    data = []

    # Inicializamos el navegador
    driver = webdriver.Chrome(options=chrome_options)
    driver.get(url)

    # Buscamos los elementos HTML que contienen el texto
    elements = driver.find_elements(By.CLASS_NAME, 'mw-content-ltr')

    try:
        # Extraemos el texto del primer elemento encontrado
        text = elements[0].text
        data.append(text)

    except:
        data.append('')

    driver.quit()

    return data
```

```python
# Scrapeamos las URLs desde la página principal
url_base = 'https://terraria.fandom.com/es' #terraria :)
url_data = [url_base] # Lista de se almacenan las URLs encontradas
n_urls = 100 # Cantidad de URLs que queremos recolectar

for url in url_data:

    # Si todavía no llegamos al límite de URLs deseadas extraemos más enlaces
    if len(url_data) < n_urls:
        get_links(url, url_base, url_data)
    else:
        break

url_data = random.sample(url_data, n_urls)
print('Cantidad de urls:', len(url_data))
```

```python
# Scrapeamos el texto de cada url
url_text = []
i = 0

for url in url_data:
  print(f"\rLinks scrapeados: {i}/{len(url_data)}", end="")
  url_text += get_text(url)
  i += 1

print(f"\rLinks scrapeados: {i}/{len(url_data)}")

url_text
```

```python
# Scrapeamos los links de salida de cada url
urls_out = []
i = 1

for url in url_data:
  links_page = []
  print(f"({i}/{len(url_data)})", end=" ")
  get_links(url, url_base, links_page)
  urls_out.append(links_page)

  i += 1
```

```python
# Construimos un dataframe y lo guardamos
df = pd.DataFrame(list(zip(url_data, url_text, urls_out)), columns=['url', 'text', 'urls_out'])
df.to_pickle('wiki_terraria.pkl')
df.head()
```
